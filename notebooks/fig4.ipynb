```python
df = pd.read_csv("/home/feity/cryoem/temp/ribo_results/classification_results.csv")
df["tomogram"] = df["tomogram"].astype(str)
df.head()
```

```python
# Draw scatter plot of accuracy, auroc, and f1 score, colored by tomogram
import seaborn as sns

# Check available columns
print(df.columns)

# Assume columns: 'tomogram', 'accuracy', 'auroc', 'f1_score' (adjust if needed)
plt.figure(figsize=(12, 4))

metrics = ['accuracy', 'auroc', 'f1']
for i, metric in enumerate(metrics):
    plt.subplot(1, 3, i+1)
    sns.scatterplot(data=df, x='tomogram', y=metric, hue='tomogram', palette='tab10', legend=False)
    plt.title(metric)
    plt.xlabel('Tomogram')
    plt.ylabel(metric)
    plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
